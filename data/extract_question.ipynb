```python
import json
import pandas as pd

file_path = r"E:\Github\LawAssistant\data\facebook_question.xlsx"

df = pd.read_excel(file_path)

questions = df["Câu hỏi"].dropna()

with open("data/facebook_questions.jsonl", "w", encoding="utf-8") as f:
    for q in questions:
        f.write(json.dumps({"question": q}, ensure_ascii=False) + "\n")
```
