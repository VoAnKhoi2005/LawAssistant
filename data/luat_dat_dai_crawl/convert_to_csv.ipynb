```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup
from urllib.parse import urljoin
from tqdm import tqdm

def normalize(text: str) -> str:
    return (
        text.strip()
        .upper()
        .replace(" ", "")
        .replace("-", "/")
    )

BASE_URL = "https://vbpl.vn/pages/vbpq-timkiem.aspx?type=0&s=0&SearchIn=Title,Title1&Keyword={keyword}"
DOMAIN = "https://vbpl.vn"

options = Options()
options.add_argument("--headless")
options.add_argument("--disable-gpu")

driver = webdriver.Chrome(options=options)

results = []

for keyword in tqdm(keywords, desc="🔎 Searching laws", ncols=80):
    url = BASE_URL.format(keyword=keyword)
    driver.get(url)

    WebDriverWait(driver, 10).until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "ul.listLaw"))
    )

    soup = BeautifulSoup(driver.page_source, "html.parser")
    list_law = soup.find("ul", class_="listLaw")

    for li in list_law.find_all("li", recursive=False):
        a_tag = li.select_one("p.title a")
        if not a_tag:
            continue

        title = a_tag.get_text(strip=True)
        href = a_tag.get("href", "")

        if normalize(keyword) in normalize(title):
            results.append({
                "keyword": keyword,
                "title": title,
                "url": urljoin(DOMAIN, href)
            })
            break  # only first match

driver.quit()

print("\nFINAL RESULTS")
for r in results:
    print(r)
```

```python
import csv

OUTPUT_FILE = "vbpl_search_results.csv"

with open(OUTPUT_FILE, mode="w", newline="", encoding="utf-8-sig") as f:
    writer = csv.DictWriter(
        f,
        fieldnames=["keyword", "title", "url"]
    )
    writer.writeheader()
    writer.writerows(results)

print(f"\nSaved {len(results)} records to {OUTPUT_FILE}")
```

```python
import pandas as pd

MAIN_CSV = r"E:\Github\uit_chatbot\graph\data\luat_dat_dai\Tổng hợp văn bản luật đất đai.csv"
VBPL_CSV = r"vbpl_search_results.csv"

def normalize(s):
    if pd.isna(s):
        return ""
    return (
        str(s)
        .strip()
        .upper()
        .replace(" ", "")
        .replace("-", "/")
    )

# ======================
# LOAD FILES
# ======================
main = pd.read_csv(MAIN_CSV, encoding="utf-8")
vbpl = pd.read_csv(VBPL_CSV, encoding="utf-8-sig")

# ======================
# PREPARE VBPL MAP
# ======================
required_cols = {"keyword", "url"}
missing = required_cols - set(vbpl.columns)
if missing:
    raise ValueError(f"VBPL CSV thiếu cột: {missing}")

vbpl["_norm_key"] = vbpl["keyword"].map(normalize)
vbpl = vbpl.dropna(subset=["url"])
vbpl = vbpl.drop_duplicates("_norm_key", keep="first")

url_map = vbpl.set_index("_norm_key")["url"].to_dict()

# ======================
# PREPARE MAIN FILE
# ======================
if "link" not in main.columns:
    main["link"] = ""

main["_norm_key"] = main["Số ký hiệu"].map(normalize)

# Count before
before_nonempty = (main["link"].fillna("").str.strip() != "").sum()

# ======================
# JOIN (SAFE WITH NaN)
# ======================
# ❗ DO NOT astype(str) BEFORE fillna
main["link"] = main["link"].fillna("")

main["link"] = main["link"].where(
    main["link"].str.strip() != "",
    main["_norm_key"].map(url_map).fillna("")
)

# Count after
after_nonempty = (main["link"].fillna("").str.strip() != "").sum()
filled_count = after_nonempty - before_nonempty

matched_keys = len(set(main["_norm_key"]) & set(url_map.keys()))
missing_keys = len(set(main["_norm_key"]) - set(url_map.keys()))

# ======================
# CLEANUP & SAVE
# ======================
main = main.drop(columns=["_norm_key"], errors="ignore")

main.to_csv(MAIN_CSV, index=False, encoding="utf-8-sig")

# ======================
# REPORT
# ======================
print(f"Rows in main: {len(main)}")
print(f"VBPL records: {len(vbpl)}")
print(f"Matched keys: {matched_keys}")
print(f"Missing keys (no vbpl url): {missing_keys}")
print(f"Newly filled links: {filled_count}")
print(f"Saved updated file to `{MAIN_CSV}`")
```

```python
from pymongo import UpdateMany
import sqlite3

def load_documents(sqlite_path: str):
    conn = sqlite3.connect(sqlite_path)
    conn.row_factory = sqlite3.Row

    rows = conn.execute("""
        SELECT id, title AS doc_title, so_hieu
        FROM documents
    """).fetchall()

    conn.close()

    return {
        row["id"]: {
            "doc_title": row["doc_title"],
            "so_hieu": row["so_hieu"]
        }
        for row in rows
    }

from pymongo import MongoClient, UpdateOne

def enrich_items(
    mongo_uri: str,
    db_name: str,
    collection_name: str,
    doc_map: dict
):
    client = MongoClient(mongo_uri)
    col = client[db_name][collection_name]

    ops = []

    for doc_id, meta in doc_map.items():
        ops.append(
            UpdateMany(
                {"doc_id": doc_id},
                {
                    "$set": {
                        "doc_title": meta["doc_title"],
                        "so_hieu": meta["so_hieu"]
                    }
                }
            )
        )

    if ops:
        result = col.bulk_write(ops)
        print("Matched:", result.matched_count)
        print("Modified:", result.modified_count)
```
