```python
import os
import re
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import time

# Create nhom_luat_giao_thong directory
os.makedirs("nhom_luat_giao_thong", exist_ok=True)

# Regex to match javascript:downloadfile('filename','path');
pattern = re.compile(r"javascript:downloadfile\('([^']+)','([^']+)'\)")

# Statistics
total_pages = len(urls)
skipped_expired = 0
skipped_no_docs = 0
downloaded_files = 0
failed_downloads = 0

print(f"Starting to process {total_pages} pages...\n")

for idx, page_url in enumerate(urls, 1):
    print(f"[{idx}/{total_pages}] Checking: {page_url}")

    try:
        response = requests.get(page_url, timeout=15)
        response.raise_for_status()

        soup = BeautifulSoup(response.text, "html.parser")

        # Check if document is expired (Hết hiệu lực toàn bộ)
        vb_info = soup.find('div', class_='vbInfo')
        if vb_info and 'Hết hiệu lực toàn bộ' in vb_info.get_text():
            print("SKIPPED: Hết hiệu lực toàn bộ\n")
            skipped_expired += 1
            time.sleep(1)
            continue

        # Find all links with downloadfile pattern
        links = soup.find_all("a", href=pattern)

        if not links:
            print("No downloadable files found.\n")
            skipped_no_docs += 1
            time.sleep(1)
            continue

        print(f"  ✓ Found {len(links)} downloadable file(s)")

        # Track if main document is downloaded
        main_doc_downloaded = False

        for a_tag in links:
            match = pattern.search(a_tag["href"])
            if match:
                filename, rel_path = match.groups()

                # Only download .doc and .docx files
                if not filename.lower().endswith(('.doc', '.docx')):
                    continue

                # Skip attachments (Phụ lục kèm theo)
                filename_lower = filename.lower()
                skip_keywords = ['phuluc', 'phu luc', 'phụ lục', 'pl ', 'phu_luc']
                if any(keyword in filename_lower for keyword in skip_keywords):
                    continue

                # Only download the first main document
                if main_doc_downloaded:
                    continue

                file_url = urljoin("https://vbpl.vn", rel_path)
                file_path = os.path.join("nhom_luat_giao_thong", filename)

                # Skip if already exists
                if os.path.exists(file_path):
                    print(f"  ✓ Already exists: {filename}")
                    main_doc_downloaded = True
                    continue

                print(f"Downloading: {filename}")
                try:
                    file_response = requests.get(file_url, timeout=20)
                    if file_response.status_code == 200:
                        with open(file_path, "wb") as f:
                            f.write(file_response.content)
                        file_size = os.path.getsize(file_path)
                        print(f"  ✓ Saved: {filename} ({file_size:,} bytes)")
                        downloaded_files += 1
                        main_doc_downloaded = True
                    else:
                        print(f"  ✗ Failed ({file_response.status_code}) for {file_url}")
                        failed_downloads += 1
                except Exception as e:
                    print(f"  ✗ Download error: {e}")
                    failed_downloads += 1

                time.sleep(0.5)

        print()
        time.sleep(1)

    except Exception as e:
        print(f"  ✗ Error fetching page: {e}\n")
        failed_downloads += 1

# Print summary
print("=" * 60)
print("SUMMARY")
print("=" * 60)
print(f"Total pages processed: {total_pages}")
print(f"Files downloaded: {downloaded_files}")
print(f"Pages skipped (expired): {skipped_expired}")
print(f"Pages skipped (no docs): {skipped_no_docs}")
print(f"Failed nhom_luat_giao_thong/errors: {failed_downloads}")
print("=" * 60)
print(f"\nDone! All documents saved in 'nhom_luat_giao_thong/' folder.")
```
