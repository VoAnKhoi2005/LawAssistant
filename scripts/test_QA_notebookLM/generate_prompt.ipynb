```python
def generate_prompt(question):
    prompt = f"""Trả lời câu hỏi dựa vào nguồn được cho và không được thêm thông tin không có trong nguồn. Câu trả lời đầy đủ thông tin và không dựa vào context câu trước đó. Nếu có nhiều trường hợp thì liệt kê tất cả. Phải có mục ở dưới cùng ghi rõ điều khoản điểm, văn bản nào dựa vào để trả lời câu hỏi, không có thì để trống. Câu hỏi: "{question}"."""
    return prompt
```

```python
import pandas as pd

excel_path = r"E:\Github\LawAssistant\scripts\test_QA_notebookLM\facebook_question.xlsx"
output_csv = r"E:\Github\LawAssistant\scripts\test_QA_notebookLM\question_prompt.csv"

df = pd.read_excel(excel_path)

rows = []
for _, row in df.iterrows():
    question = row["Câu hỏi"]
    if pd.isna(question):
        continue

    prompt = generate_prompt(question)

    rows.append({
        "question": question,
        "prompt": prompt
    })

out_df = pd.DataFrame(rows)
out_df.to_csv(output_csv, index=False, encoding="utf-8-sig")
```
